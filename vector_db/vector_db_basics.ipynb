```python
import chromadb

client = chromadb.Client()
```

```python
collections = client.create_collection("news_articles")
```

```python
collections.add(
    ids=["id1","id2","id3","id4"],
    documents=[
        "Apple is leading the smart phone game with iphone sales up by 35%",
        "Tesla booked a minor profit of 1 billion $ in Q2",
        "Apples are high in fiber, vitamin C, and various antioxidants",
        "SpaceX got NASA contract worth 10 billion $"
    ]
)
```

Above it wil create embeddings for each of these sentences and it will store that into local chroma DB. chroma uses all-MiniLM-L6-v2 model from sentence-tranfomers by default for embedding

```python
#Fetch documents + embeddings
data = collections.get(
    include=["documents","embeddings","metadatas"]
)
#print embeddings
for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}")
    print("Text:",data["documents"][i])
    print("Embedding length:",len(emb))
    print("FIrst 10 values:",emb[:10])
```

```python
results=collections.query(
    query_texts=["This is a query related to Elon Musk"],
    n_results=2
)
print(results)
```
